```python
def chunk(item, size):
    for i in range(0,len(item),size):
        yield item[i:i+size]

print(list(chunk([1,2,3,4,5,6,7,8,9],4)))
```

```python
files = dict(zip(["a.txt","b.txt","c.txt"],[["hello","world"],["foo"],["bar","baz"]]))

def read_file(name):

    for s in files[name]:
        str_ = name + ": " + s
        yield str_
    
def read_all(names):
    
    for name in names:
        if name not in files:
            print(f"{name} 파일은 존재하지 않습니다. 건너뛰겠습니다.")
            print()
            continue 
        yield from read_file(name)

for line in read_all(["a.txt","r.txt","b.txt","q.txt","c.txt"]):
    print(line)
    print()
```

```python
def filter_even(nums):
    deserted_odds_count = 0
    
    for d in nums:
        if d%2 == 0:
            yield d
        else:
            deserted_odds_count+=1
            
    return deserted_odds_count

def report(nums):
    N = yield from filter_even(nums)
    yield f"홀수 {N}개 제외"

# 3항 연산자 + 제너레이터 컴프리헨션
squared = (x**2 if isinstance(x,int) else x for x in report(list(range(1,7))))

for x in squared:
    print(x)

# 여기서 odds_count, result_list 를 모두 yield 로 내보내면 구분이 안되서 구분하기 위해서 이렇게 전달한건가 ? -> 맞음. 자주 쓰는 방식임
        
```
